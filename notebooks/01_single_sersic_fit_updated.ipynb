# 01 — Single Sérsic mock

Minimal notebook to generate and display a synthetic Sérsic galaxy.

```python
import matplotlib.pyplot as plt

from baymorph import SyntheticSersic
```

```python
galaxy = SyntheticSersic(
    flux=10000.0,
    re=10.0,
    n=4.0,
    q=0.7,
    theta=0.5,
    x0=50.0,
    y0=50.0,
)
```

```python
image = galaxy.render(
    shape=(100, 100),
)
```

```python
plt.figure(figsize=(6, 6))

plt.imshow(
    image,
    origin="lower",
)

plt.colorbar(label="Flux")
plt.title("Synthetic Sérsic galaxy")
plt.show()
```

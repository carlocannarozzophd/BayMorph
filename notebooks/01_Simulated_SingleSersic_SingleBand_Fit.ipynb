# 01 — Simulated Single-Sérsic Single-Observation Fit

Minimal end-to-end BayMorph validation on a controlled simulated galaxy.

The simulation truth, morphology model, priors, likelihood, and sampler
configuration are stored in the example YAML.


```python
from pathlib import Path

from baymorph import BayMorphRunner
from baymorph.pipeline import run_simulation_from_config


ROOT = (
    Path.cwd().parent
    if Path.cwd().name == "notebooks"
    else Path.cwd()
)

CONFIG = ROOT / "configs/simulated_galaxy.yaml"
```

## Simulate and fit

BayMorph generates the mock observation and then runs the same production
pipeline used for observed data.


```python
run_simulation_from_config(CONFIG)

runner = BayMorphRunner(
    CONFIG,
    verbose=True,
)

result = runner.run(
    diagnostics=True,
)
```

## Summary


```python
{
    "truth": runner.config["simulation"]["truth"],
    "fit": result.summary(),
}
```

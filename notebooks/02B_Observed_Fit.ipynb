# Fitting an Observed Galaxy

This notebook fits the models of the configuration written by
`02A_Observed_Assessment_and_Config.ipynb`. 02A already prepared the image
(cutout, background, noise and mask) and built the empirical PSF; here these
stages load the accepted products and show them. The notebook then fits each
model, shows its posterior and residuals, compares the evidence of the
models and writes the diagnostics.

Every cell can be run again. Accepted products are immutable, so a stage that
finds a completed product loads it and says so instead of rebuilding it, and
the fit resumes its Nautilus checkpoint. After an interruption, rerun the
notebook from the top: it continues where it stopped.

Changing a declaration is a different scientific problem. Go back to 02A and
choose a new `analysis_name`; products are never rewritten in place.

```python
import numpy as np
import yaml
from astropy.table import Table
from IPython.display import display

from baymorph import BayMorphRunner
from baymorph.plotting import (
    corner_plot,
    plot_log_data_model_residual,
    plot_log_image,
)
```

## Configuration

`config_path` is the YAML written by 02A, resolved from the project root. It
declares every prior the fit samples, for each model. The sampler settings
below add the defaults BayMorph filled in.

```python
config_path = "configs/galaxy_r.yaml"

runner = BayMorphRunner(config_path)
object_name = runner.config["object"]["name"]
models = runner.config["fit"]["models"]

print(f"Configuration: {runner.config_path}")
print(f"Run directory: {runner.run_dir}")
print(f"Models: {', '.join(models)}")
print()
print(runner.config_path.read_text(encoding="utf-8"))
print("Resolved sampler settings:")
print(yaml.safe_dump(runner.config["fit"]["sampler"], sort_keys=False))
```

## Image preparation

`prepare_image()` loads the cutout 02A prepared, with its sky removed, its
noise scale and its fitting mask, or builds them if they are missing. Masked
pixels (neighbouring sources, deblended fragments, artefacts) do not enter
the likelihood. A fixed cutout too small to contain the sky plateau around
the galaxy is rejected; the remedy is a larger cutout, declared in 02A under
a new `analysis_name`.

Look at the mask before fitting: large masked regions inside the galaxy, or
unmasked artefacts, change the scientific problem. The status line says
whether this stage built the products or reused accepted ones.

```python
image = runner.prepare_image()

height, width = image.prepared_data.shape

print(f"Cutout: {width} x {height} pixels ({image.cutout_mode})")
print(f"Pixels excluded by the fitting mask: {image.fitting_mask.mean():.1%}")
print(f"Noise scale: {np.median(image.noise):.4g}")
```

```python
figure, _ = plot_log_image(
    image.prepared_data,
    title=f"{object_name}: prepared cutout",
    figsize=(15, 15),
    show=True,
)

figure, _ = plot_log_image(
    np.where(image.fitting_mask, np.nan, image.prepared_data),
    title=f"{object_name}: pixels used by the likelihood",
    figsize=(15, 15),
    show=True,
)
```

### Sky gradient

Preparation measures the noise scale about a plane fitted to the sky and
reports the plane's gradient. The gradient is a diagnostic and never changes
the background model. A gradient significant at several sigma indicates that
`model: plane` is warranted, which is a new declaration in 02A.

A sky discontinuity, such as the edge of a coadd footprint, is modelled by
neither `constant` nor `plane`. A smaller fixed cutout can exclude it.

```python
gradient = image.sky_gradient

if gradient is None:
    print("No sky gradient was measured.")
else:
    for axis in ("x", "y"):
        value = gradient[f"gradient_{axis}"]
        error = gradient[f"gradient_{axis}_error"]
        significance = gradient[f"gradient_{axis}_significance"]

        print(
            f"gradient_{axis}: {value:.3g} +/- {error:.2g} per pixel "
            f"({significance:.1f} sigma)"
        )
```

## Empirical PSF

BayMorph selects isolated stars whose peak lies below `psf_max_peak` and
combines them into one PSF kernel. It rejects a kernel that is not finite,
non-negative, normalized and centred. 02A built the PSF; these calls load
it.

```python
runner.select_psf_candidates()
psf = runner.build_psf()

figure, _ = plot_log_image(
    psf,
    title=f"{object_name}: empirical PSF",
    figsize=(15, 15),
    show=True,
)
```

## Fit

`fit()` runs Nautilus for each model in turn, with its declared priors and
the same likelihood. This is the slow cell: its cost grows with the number of
pixels used by the likelihood and with the number of parameters, so a model
with several components takes longer than a single Sérsic. Each model keeps
its own checkpoint, so an interrupted run resumes the model it stopped in.

```python
result = runner.fit()

# One model gives its products; several give a mapping from model to products.
fits = result if isinstance(result, dict) else {models[0]: result}
```

## Posterior

For each model, the median and `q16`-`q84` credible interval of each
parameter. `x0` and `y0` are in cutout pixel coordinates, and
`background_level` is the residual background after preparation, not the
sky level.

```python
for model_name, fit in fits.items():
    inference = fit.posterior

    posterior_table = Table(
        rows=[
            {
                "parameter": name,
                **inference.summary(name),
            }
            for name in inference.parameter_names
        ]
    )

    print(model_name)
    display(posterior_table)
```

## Residuals

Data, best-fitting model and residuals over the pixels used by the
likelihood, for each model. The normalized residual divides by the scale of
the noise model at the best sample, which adds the source Poisson variance
to the sky noise when the configuration declares an effective gain (AD-047).
It is not whitened, so neighbouring pixels keep the correlation of the sky.
Structure in the residuals shows where the morphology model or the
observation model falls short.

```python
for model_name, fit in fits.items():
    used = np.isfinite(fit.norm_residual)

    residual_figure = plot_log_data_model_residual(
        np.where(used, fit.best_model_image + fit.residual, np.nan),
        fit.best_model_image,
        norm_residual=fit.norm_residual,
        title=f"{object_name}: {model_name}",
        show=True,
    )
```

## Evidence

Each fit reports the logarithm of its Bayesian evidence, `logZ`, with the
uncertainty of the sampler. All models were fitted to the same prepared
image, mask, PSF and likelihood, so the differences of `logZ` compare them.
The evidence depends on the prior of each model: it compares the models as
declared, priors included. On the scale of Kass & Raftery (1995), a
difference above 1 is positive, above 3 strong and above 5 very strong; a
difference within the uncertainties decides nothing. The evidence does not
replace the residuals: the preferred model can still fail to describe the
image.

```python
evidences = {
    model_name: fit.posterior.logz
    for model_name, fit in fits.items()
    if fit.posterior.logz is not None
}

if not evidences:
    print("Evidence: unavailable (posterior-only backend)")
else:
    best_logz = max(evidences.values())

    evidence_table = Table(
        rows=[
            {
                "model": model_name,
                "parameters": len(fits[model_name].posterior.parameter_names),
                "logZ": logz,
                "logZ error": fits[model_name].posterior.logz_error,
                "logZ - best": logz - best_logz,
                "reduced chi2": fits[model_name].reduced_chi2,
            }
            for model_name, logz in evidences.items()
        ]
    )

    display(evidence_table)
```

## Diagnostics

`diagnostics()` writes the corner plots, residual maps and radial profiles of
every model as PDF files in the run directory. The cell below also draws the
corner plot of each model inline.

```python
runner.diagnostics()

print(f"Diagnostics written to: {runner.run_dir}")
```

```python
for model_name, fit in fits.items():
    corner_figure, _, _ = corner_plot(
        fit.posterior.equal_weight_samples.T,
        parameter_labels=fit.posterior.parameter_names,
        title=f"{object_name}: {model_name}",
    )
```

## Summary and limitations

The reduced chi-squared is the quadratic form of the noise model of AD-047
divided by the degrees of freedom. That model adds to the sky noise the
galaxy's own Poisson noise, through the effective gain the configuration
declares, and the correlation of the sky. A reduced chi-squared well above 1
means that the model does not describe the image within that noise: a misfit
of the morphology or of the background, which no noise model removes, or a
wrong noise model, for example a wrong gain. If the configuration declares
the source variance absent, the noise in the centre of a bright galaxy is
underestimated and the credible intervals are too narrow.
`prior_boundary_diagnostics` in the summary of each model flags posteriors
that press against a declared prior bound.

```python
for model_name, fit in fits.items():
    inference = fit.posterior

    print(model_name)
    print(f"Backend: {inference.provenance.get('backend', 'unknown')}")
    print(f"Reduced chi2: {fit.reduced_chi2:.4f} (ndof = {fit.ndof})")

    if inference.logz is None:
        print("Evidence: unavailable (posterior-only backend)")
    else:
        print(f"logZ: {inference.logz:.4f}")

    display(fit.summary())
```

# Learn BayMorph

A minimal end-to-end BayMorph analysis using the recommended configuration-driven workflow.

The tutorial creates a synthetic galaxy, fits it, and generates the standard diagnostics. BayMorph handles the scientific workflow internally.

## Setup

```python
from pathlib import Path
import os

from baymorph import BayMorphRunner
from baymorph.pipeline import run_simulation_from_config


if Path.cwd().name == "notebooks":
    os.chdir(Path.cwd().parent)

CONFIG = Path("configs/learn_baymorph.yaml")
```

## Create the tutorial dataset

```python
run_simulation_from_config(CONFIG)
```

## Run BayMorph

```python
runner = BayMorphRunner(CONFIG)
fit = runner.run(diagnostics=True)
```

## Results

```python
{
    "model": fit.model_name,
    "statistics": fit.statistics,
    "best_parameters": fit.best_parameters,
    "output_directory": str(runner.output_dir),
}
```

## Done

The resolved configuration, prepared image, posterior products, best-fitting model, residual maps, corner plot, and radial-profile diagnostics are available in the output directory shown above.

For observed data, point the same runner workflow to a configuration containing the science image and PSF settings.
